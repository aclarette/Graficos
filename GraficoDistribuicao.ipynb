```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Certifique-se de usar o caminho correto para o seu arquivo
tb = pd.read_csv("graficosCSV/dados_distribuicao.csv")

fig, ax = plt.subplots(figsize=(8, 5))

# Plot do histograma com KDE utilizando a coluna 'idade'
sns.histplot(
    data=tb,
    x="idade",
    kde=True,
    bins=15,
    color="#4c72b0",
    ax=ax,
)

# Ajuste dos rótulos e título
ax.set_title("Distribuição da Idade dos Clientes", fontsize=14, fontweight="bold")
ax.set_xlabel("Idade (anos)", fontsize=11)
ax.set_ylabel("Frequência", fontsize=11)

plt.tight_layout()
plt.show()
```
